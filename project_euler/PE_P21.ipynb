```python
def d(x):
    factors = set()
    factors.add(1)
    i = 2
    while i * i <= x:
        if x % i == 0:
            factors.add(i)
            factors.add(x // i)
        i += 1
    return sum(factors)

print(d(220))
```

```python
seen = set()
paired = set()

for i in range(10000):
    if i not in seen:
        d_val = d(i)
        d_val_d_val = d(d_val)
        if d_val_d_val == i and d_val != i:
            print(d_val, i)
            paired.add(d_val)
            paired.add(i)
        seen.add(i)
        seen.add(d_val)

print(paired)

print(sum(paired))
```
