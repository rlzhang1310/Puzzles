```python
# 3 sides can be generated with m^2 + n^2, m^2 - n^2, 2mn, where m > n

from collections import defaultdict
```

```python
seen_triangles = set()
perimeter_counter = defaultdict(int)
for m in range(2, 25):
    for n in range(1, m):
        a = m**2 + n**2
        b = m**2 - n**2
        c = 2*m*n
        perim = a+b+c
        if perim <= 1000:
            side_lst = [a, b, c]
            side_lst.sort()
            side_lst_tuple = tuple(side_lst)
            multiplier = 1
            while multiplier * perim <= 1000:
                new_tuple = (side_lst_tuple[0] * multiplier, side_lst_tuple[1] * multiplier, side_lst_tuple[2] * multiplier)
                multiplier += 1
                # if perim == 120:
                #     print(new_tuple)
                if new_tuple not in seen_triangles:
                    perimeter_counter[perim * multiplier] += 1
                    seen_triangles.add(new_tuple)


print(seen_triangles)
print(perimeter_counter)
print(perimeter_counter[120])
```

```python
max_v = 0
max_k = 0
for k,v in perimeter_counter.items():
    if v >max_v:
        max_v = v
        max_k = k

print(max_k, max_v)
```
