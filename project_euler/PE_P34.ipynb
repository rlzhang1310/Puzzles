```python
from math import factorial

factorial_dict = dict()
for i in range(0, 10):
    factorial_dict[str(i)] = factorial(i)

print(factorial_dict)
```

```python
def checkNum(x):
    str_x = str(x)
    val = 0
    for d in str_x:
        val += factorial_dict[d]
        if x < val:
            return False
        
    if x == val:
        return True
    return False
```

```python
valid_nums = set()

for i in range(10, 362880 * 25):
    if checkNum(i):
        valid_nums.add(i)

print(valid_nums)
print(sum(valid_nums))
```
