```python
from sympy import isprime
```

```python
truncatable = set()
#  generate right to left, and verify with left to right
# right to left are all just prime that aren't odd: 1, 3, 7, 9
# 2 digit ones we can think of: 23, 37, 53, 73
# we are missing 7 more

def check_left_to_right(n):
    val = n
    digits = len(str(n))
    while digits > 0:
        val = val % 10**digits
        digits -= 1
        # print(val)
        if not isprime(val):
            return False
        
    return True

check_left_to_right(23)
```

```python
lst = [2, 3, 5, 7]

while len(truncatable) < 11:
    new_lst = []
    for item in lst:
        for i in [1, 3, 7, 9]:
            new_val = item*10 + i
            if isprime(new_val):
                new_lst.append(new_val)
            if check_left_to_right(new_val):
                truncatable.add(new_val)
    lst = new_lst
    print(lst)
```

```python
print(truncatable)
print(len(truncatable))
print(sum(truncatable))
```
