```python
# div by 17
n = 1
valid = set()
while n * 17 <= 1000:
    curr = n * 17
    if len(set(list(str(curr)))) == 3:
        valid.add(str(curr))
    elif curr < 100 and len(set(list(str(curr)))) == 2:
        valid.add('0' + str(curr))
    n += 1
```
