```python
from math import gcd

for i in range(10,99):
    for j in range(i + 1,99):
        gcd_i_j = gcd(i, j)
        if gcd_i_j != 1 and (i // 10 == j // 10 or i % 10 == j // 10 or i // 10 == j % 10):
            for k in range(1, gcd_i_j):
                i_fact = i // gcd_i_j * k
                j_fact = j // gcd_i_j * k
                if str(i_fact) in str(i) and str(j_fact) in str(j):
                    print(i, j)
```
