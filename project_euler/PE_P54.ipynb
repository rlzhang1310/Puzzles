```python
mp = dict()
lychrel = set()
def lychrel_checker(n, count):
    # print(n, count)
    if count > 50:
        return True
    reverse_n = int(str(n)[::-1])
    if reverse_n == n and count != 0:
        return False
    curr_sum = n + reverse_n
    if curr_sum in mp:
        return mp[curr_sum]
    curr_sol = lychrel_checker(curr_sum, count+1)
    mp[curr_sum] = curr_sol
    return curr_sol

for i in range(1, 10001):
    if lychrel_checker(i, 0):
        lychrel.add(i)


print(lychrel)
print(len(lychrel))
```
