```python
counter = 0
dp = [1, 1]
for i in range(2, 101):
    new_dp = [1]
    for i in range(len(dp) - 1):
        curr_sum = dp[i] + dp[i + 1]
        if dp[i] == 0 or dp[i + 1] == 0 or curr_sum > 1000000:
            new_dp.append(0)
            counter += 1
        else:
            new_dp.append(curr_sum)
    new_dp.append(1)
    dp = new_dp

print(counter)

```
