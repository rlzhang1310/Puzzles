```python
dp = []
dp.append([75])

for i in range(1, len(pyramid)):
    cur_dp = [0] * len(pyramid[i])
    for j in range(len(pyramid[i])):
        if j == 0:
            cur_dp[j] = dp[i - 1][0]
        elif j == len(pyramid[i]) - 1:
            cur_dp[j] = dp[i - 1][len(pyramid[i]) - 2]
        else:
            cur_dp[j] = max(dp[i - 1][j - 1], dp[i - 1][j])
        cur_dp[j] += pyramid[i][j]
    dp.append(cur_dp)


print(max(dp[len(pyramid) - 1]))
print(dp)
```
