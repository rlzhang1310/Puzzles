```python
digits = [1]
```

```python
for j in range(1000):
    carry = 0
    for i in range(len(digits)):
        doubled = digits[i] * 2 + carry
        if doubled >= 10:
            carry = 1
            doubled -= 10
        else:
            carry = 0
        digits[i] = doubled
    if carry == 1:
        digits.append(1)
    print(j + 1, digits)


print(sum(digits))
```
