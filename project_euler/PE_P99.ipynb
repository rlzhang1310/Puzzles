```python
def exponent_size(a, b):
    dp = [1]
    for _ in range(b):
        for i in range(len(dp)):
            dp[i] *= a
        carry = 0
        for i in range(len(dp)):
            dp[i] += carry
            carry = 0
            if dp[i] > 10**10:
                carry = dp[i] // 10**10
                dp[i] = dp[i] % 10**10
        if carry > 0:
            dp.append(carry)
        curr_sum = 0
        for i in dp:
            for d in str(i):
                curr_sum += int(d)
    
    return (len(dp) - 1) * 10 + len(dp[-1]), dp[-1]

def function():
    largest_length = 0
    largest_last = 0
    la = 0
    lb = 0
    lines = 0
    with open("./txt_files/0099_base_exp.txt", 'r') as file:
        for line in file:
            print(line)
            if lines > 20:
                print(la, lb)
                return
            a, b = line.strip().split(',')
            length, last = exponent_size(int(a), int(b))
            if length > largest_length or (length == largest_length and last > largest_last):
                largest_length = length
                largest_last = largest_last
                la = a
                lb = b

function()
```
