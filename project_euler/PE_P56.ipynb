```python
largest_digit_sum = 0
la = 1
lb = 1
for a in range(2, 100):
    dp = [1]
    for b in range(1, 100):
        for i in range(len(dp)):
            dp[i] *= a
        carry = 0
        for i in range(len(dp)):
            dp[i] += carry
            carry = 0
            if dp[i] > 10**10:
                carry = dp[i] // 10**10
                dp[i] = dp[i] % 10**10
        if carry > 0:
            dp.append(carry)
        curr_sum = 0
        for i in dp:
            for d in str(i):
                curr_sum += int(d)
        if curr_sum > largest_digit_sum:
            largest_digit_sum = curr_sum
            la = a
            lb = b
        print(a,b,curr_sum, dp)

```
