```python
digit_count = 0
digits_looking_for = [1, 10, 100, 1000, 10000, 100000, 1000000]


for i in range(1, 500000):
    to_add = len(str(i))
    for d in digits_looking_for:
        if digit_count < d and digit_count + to_add >= d:
            print(d, digit_count, i)
    digit_count += to_add
    
```
