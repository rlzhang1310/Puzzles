```python
# let a, b, c be represented as
# a = 2mn, b = m^2 - n^2, c = m^2 + n^2

# m(m + n) = 500

m = 1
while m * m <= 500:
    if 500 % m == 0:
        m_plus_n = 500 // m
        n = m_plus_n - m
        a = 2*m*n
        b = m**2 - n**2
        c = m**2 + n**2
        print(m, n)
        # if a > 0 and b > 0 and c > 0:
        print(a, b, c)
    m += 1
```
