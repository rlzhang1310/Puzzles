```python
import itertools
from sympy import isprime


def createPossiblePermutations(n):
    lst_of_nums = range(1, n + 1)
    lst_of_nums = [str(x) for x in lst_of_nums]
    perms = list(itertools.permutations(lst_of_nums))
    # print(type(perms[0]))
    int_perms = [int("".join(x)) for x in perms]
    # print(int_perms)
    return int_perms


createPossiblePermutations(3)
```

```python
max_v = 1
for i in range(2,10):
    if i % 3 != 1:
        continue
    perms_to_check = createPossiblePermutations(i)
    for n in perms_to_check:
        if isprime(n):
            max_v = max(max_v, n)
            # print(n)


print(max_v)
```
