```python
from itertools import permutations
lst = [x + 1 for x in range(9)]
print(lst)
combos = list(permutations(lst))
print(len(combos))
```

```python
works = set()
for combo in combos:
    for i in range(1, 8):
        for j in range(i + 1, 9):
            x1 = combo[:i]
            x2 = combo[i:j]
            x3 = combo[j:]
            if len(x3) < len(x2) or len(x3) < len(x2) or len(x3) > len(x1) + len(x2):
                continue
            else:
                num_x1 = int("".join(str(i) for i in x1))
                num_x2 = int("".join(str(i) for i in x2))
                num_x3 = int("".join(str(i) for i in x3))
                if num_x1 * num_x2 == num_x3:
                    print(num_x1, num_x2, num_x3)
                    works.add(num_x3)

print(works)
```

```python
print(sum(works))
```
