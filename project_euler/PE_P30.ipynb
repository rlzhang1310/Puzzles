```python
fifth_pow_dict = dict()
for i in range(0, 10):
    fifth_pow_dict[str(i)] = i**5

print(fifth_pow_dict)
```

```python
def checkNum(x):
    str_x = str(x)
    val = 0
    for d in str_x:
        val += fifth_pow_dict[d]
        if x < val:
            return False
        
    if x == val:
        return True
    return False
```

```python
valid_nums = set()

for i in range(10, 59049 * 5):
    if checkNum(i):
        valid_nums.add(i)

print(valid_nums)
print(sum(valid_nums))
```
