```python
numbers_left = []
n = 10
for i in range(n):
    numbers_left.append(str(i))
print(numbers_left)

def helper(count):
    permutations_left = count - 1
    if count == 0 or len(numbers_left) == 1:
        return "".join(numbers_left)
    if count == 0 or len(numbers_left) == 0:
        return "".join(numbers_left)
    n = len(numbers_left)
    if permutations_left == math.factorial(n):
        return "".join(reversed(numbers_left))
    print(permutations_left, math.factorial(n - 1), numbers_left)
    divisor, mod = permutations_left // math.factorial(n - 1), permutations_left % math.factorial(n - 1)
    letter_to_add = numbers_left[divisor]
    numbers_left.remove(letter_to_add)
    return letter_to_add + helper(mod + 1)

helper(1000000)
```
