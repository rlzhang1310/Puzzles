```python
from sympy import isprime

def findnext():
    primes = set()
    non_primes = set()
    n=9
    while 1==1:
        if isprime(n):
            primes.add(n)
            n += 2
            continue
        else:
            non_primes.add(n)
        print(n)
        i = 1
        while 1==1:
            print(n, i)
            curr = n - 2*i*i
            if curr < 0:
                print(n)
                return n
            elif curr not in primes and curr not in non_primes:
                if isprime(curr):
                    primes.add(curr)
                    break
                else:
                    non_primes.add(curr)
            elif curr in primes:
                break
            i += 1
        n += 2
    return -1
        
        
```
