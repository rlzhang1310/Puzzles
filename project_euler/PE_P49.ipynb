```python
from sympy import isprime

for i in range(1000, 9999+1):
    if not isprime(i):
        continue
    for inc in range(10, 4500):
        if not isprime(i + inc) or not set(str(i)) == set(str(i + inc)):
            continue
        if set(str(i)) == set(str(i + 2*inc)):
            # print(i, i+inc, i+2*inc)
            if isprime(i+inc) and isprime(i+2*inc):
                print("THIS ONE WORKS: ", i, i+inc, i+2*inc)



        
```
