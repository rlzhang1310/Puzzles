```python
def generate_pentagonal_number(n):
    pn = n * (3 * n - 1) // 2
    return pn
```

```python
pent_set = set()
for j in range(1, 100000):
    pj = generate_pentagonal_number(j)
    pent_set.add(pj)

max_pent = max(pent_set)
```

```python
def find_best_j():
    for j in range(1, 1200):
        print("none for ", j)
        for n in range(1, 100000):
            diff = 3*j*n + (3*j*j-j) // 2
            sum = 3*n**2 - n + diff
            if diff in pent_set and sum in pent_set:
                print(n, j, diff, sum)
                return j

            if sum > max_pent:
                break

    return 1000


find_best_j()
```
