```python
triangle_numbers = set()
pentagonal_numbers = set()
hexagonal_numbers = set()

def triangle(n):
    return n * (n + 1) // 2

def pentagon(n):
    return n * (3 *n - 1) // 2

def hexagon(n):
    return n * (2 *n - 1)


for i in range(1, 100000):
    triangle_numbers.add(triangle(i))
    pentagonal_numbers.add(pentagon(i))
    hexagonal_numbers.add(hexagon(i))


for i in range(143, 100000):
    curr = hexagon(i)
    # print(i)
    if curr in triangle_numbers and curr in pentagonal_numbers:
        print(i)
        print(curr)

```
