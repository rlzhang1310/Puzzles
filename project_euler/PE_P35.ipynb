```python
from sympy import isprime
```

```python
def generate_candidates():
    candidates = set()
    candidates.add(1)
    candidates.add(3)
    candidates.add(7)
    candidates.add(9)
    for digits in range(5):
        to_add = set()
        for i in [1, 3, 7, 9]:
            for candidate in candidates:
                to_add.add(candidate * 10 + i)
        print(to_add)
        candidates = candidates.union(to_add)

        print(digits, candidates)
    return candidates

candidates = generate_candidates()
# print(generate_candidates())
candidates.remove(1)
candidates.remove(3)
candidates.remove(7)
candidates.remove(9)
```

```python
print(max(candidates))
```

```python
def createRotates(n):
    lst = []
    str_n = str(n)
    for i in range(len(str_n)):
        rotated = str_n[i:] + str_n[0:i]
        lst.append(int(rotated))
    return lst

createRotates(3137339393)
```

```python
circular = set()

for candidate in candidates:
    if candidate not in circular:
        # print(candidate)
        is_circular = True
        for rotates in createRotates(candidate):
            if not isprime(rotates):
                is_circular = False
        if is_circular:
            for rotates in createRotates(candidate):
                circular.add(rotates)
                


print(circular)
print(len(circular))
```
