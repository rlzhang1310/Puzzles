```python
from sympy import isprime

def get_prime_factors(n):
    prime_set = set()
    if isprime(n):
        prime_set.add(n)
        prime_set.add(1)
        return prime_set
    i = 2
    while n != 1:
        if isprime(n):
            prime_set.add(n)
            return prime_set
        curr = 1
        while n % i == 0:
            curr *= i
            n //= i
        if curr != 1:
            prime_set.add(curr)
        i += 1

    return prime_set

print(get_prime_factors(2))
```
