```python
def cycle_counter(n):
    remainder = 10
    remainder_lst = []
    remainder_set = set()
    while remainder != 0 and remainder not in remainder_set:
        # print(remainder, remainder_lst, remainder_set)
        remainder_set.add(remainder)
        if remainder < n:
            remainder *= 10
        else:
            left = remainder % n
            remainder = left * 10
        remainder_lst.append(remainder)
    # print(remainder, remainder_lst, remainder_set)
    if remainder == 0:
        return 0
    else:
        return remainder_lst.index(remainder) + 1
```

```python
max_recurrence = 0
recurrence_num = 0
for i in range(1, 1000):
    cycles = cycle_counter(i)
    if cycles > max_recurrence:
        max_recurrence = cycles
        recurrence_num = i

print(max_recurrence, recurrence_num)
```
