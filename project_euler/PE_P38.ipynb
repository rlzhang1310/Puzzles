```python
def contains_1_to_9(n):
    digits = set()
    for i in n:
        digits.add(i)
    if "0" not in digits and len(digits) == 9 and len(n) == 9:
        return True
    else:
        return False

# print(contains_1_to_9('123456789'))
# print(contains_1_to_9('987654321'))
# print(contains_1_to_9('123840951'))
# print(contains_1_to_9('987654320'))
```

```python
# testing 9xxx times 1 concat 9xxx times 2
curr_max = 123456789
base = 9000
for i in range(9000, 9999):
    concat = str(i) + str(i * 2)
    if contains_1_to_9(concat):
        print(i)
        curr_max = max(curr_max, int(concat))
        print(curr_max)
```
