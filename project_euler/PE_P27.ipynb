```python
def isPrime(n):
    i = 2
    if n <= 1:
        return False
    while i * i <= n:
        if n % i == 0:
            return False
        i += 1
    return True

isPrime(4)
```

```python
# create primes from 0 to 1000*1000*2 + 1000
prime_set = set()
prime_set_b = set()
for i in range(2001001):
    if isPrime(i):
        prime_set.add(i)
        if i <= 1000:
            prime_set_b.add(i)

print(prime_set_b)
```

```python
def calculate_n(a, b):
    n = 0
    vals = 0
    while n * n + n * a + b in prime_set:
        vals += 1
        n += 1
    return vals
```

```python
best_pair = (1, 41)
best_n = 40
for b in prime_set_b:
    for a in range(-999, 1001, 2):
        curr_n = calculate_n(a, b)
        if curr_n > best_n:
            best_pair = (a, b)
            best_n = curr_n

print(best_n, best_pair)
print(best_pair[0] * best_pair[1])
```
