```python
# for every base of n from 2-100, we calculate the highest it goes to
#  skip the ones that can be reached by raising other to the power of
# aka remove 4, 8, 16, 32, 64, 9, 27, 81, 25, 36, 49

from collections import defaultdict
```

```python
base_dict = defaultdict(set)

a_highest = 100
b_highest = 100

for a in range(2, a_highest + 1):
    for b in range(2, b_highest + 1):
        base_dict[a].add(b)


for i in range(2, a_highest):
    power = 2
    while i**power in base_dict:
        new_set = set([x * power for x in base_dict[i**power]])
        base_dict[i] = base_dict[i].union(new_set)
        del base_dict[i**power]
        power += 1


total = 0
for item in base_dict:
    total += len(base_dict[item])

print(base_dict)
print(total)

```
